# Topics: Merge BERT Output

Merges BERT inference results from `work_topics_lm_output` into `work_topics`. Enriches raw topic IDs with display names and hierarchy (subfield/field/domain), keeps top 3 by score.

- **WHEN MATCHED**: overwrites existing topics (BERT takes priority over backfill)
- **WHEN NOT MATCHED**: inserts new predictions

```sql
MERGE INTO openalex.works.work_topics AS target
USING (
  WITH topics_metadata AS (
    SELECT
      topic_id,
      t.display_name,
      NAMED_STRUCT(
        'id', concat('https://openalex.org/subfields/', s.subfield_id),
        'display_name', s.display_name
      ) AS subfield,
      NAMED_STRUCT(
        'id', concat('https://openalex.org/fields/', f.field_id),
        'display_name', f.display_name
      ) AS field,
      NAMED_STRUCT(
        'id', concat('https://openalex.org/domains/', d.domain_id),
        'display_name', d.display_name
      ) AS domain
    FROM openalex.common.topics t
    JOIN openalex.common.subfields s USING (subfield_id)
    JOIN openalex.common.fields f USING (field_id)
    JOIN openalex.common.domains d USING (domain_id)
  ),

  -- oxjob #859: work_topics_lm_output is append-only and is also the #713 attempts
  -- ledger, so a re-scored work keeps every prior prediction. Aggregating across all of
  -- them let a superseded prediction survive in the top 3 (measured: 12.0% of a 10k
  -- re-scored sample kept the stale topic, 1.5% kept it as primary). Take only the most
  -- recent prediction per work.
  latest_lm_output AS (
    SELECT work_id, lm_topics, source, created_timestamp
    FROM (
      SELECT *,
             ROW_NUMBER() OVER (
               PARTITION BY work_id
               ORDER BY created_timestamp DESC,
                        CASE WHEN lm_topics IS NULL THEN 1 ELSE 0 END
             ) AS rn
      FROM openalex.works.work_topics_lm_output
    )
    WHERE rn = 1
  ),

  lm_output_exploded AS (
    SELECT 
      work_id,
      explode(lm_topics) AS result,
      source,
      created_timestamp
    FROM latest_lm_output
  )

  SELECT
    work_id,
    slice(array_sort(
      array_agg(
        NAMED_STRUCT(
          'id', concat('https://openalex.org/T', result.topic_id),
          'display_name', tm.display_name,
          'score', ROUND(result.score, 4),
          'subfield', tm.subfield,
          'field', tm.field,
          'domain', tm.domain
        )
      ),
      (left, right) -> CASE
        WHEN left.score > right.score THEN -1
        WHEN left.score < right.score THEN 1
        WHEN left.id < right.id THEN -1
        WHEN left.id > right.id THEN 1
        ELSE 0
      END
    ), 1, 3) AS topics,
    first(wt.source) AS source,
    max(wt.created_timestamp) AS created_datetime,
    max(wt.created_timestamp) AS updated_datetime
  FROM lm_output_exploded wt
  JOIN topics_metadata tm ON tm.topic_id = result.topic_id
  GROUP BY work_id
) AS source
ON target.work_id = source.work_id

WHEN MATCHED THEN UPDATE SET
  target.topics = source.topics,
  target.source = source.source,
  target.updated_datetime = source.updated_datetime

WHEN NOT MATCHED THEN INSERT (
  work_id,
  topics,
  source,
  created_datetime,
  updated_datetime
) VALUES (
  source.work_id,
  source.topics,
  source.source,
  source.created_datetime,
  source.updated_datetime
);
```
