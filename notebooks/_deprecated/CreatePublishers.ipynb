# DEPRECATED (2026-07-06)

`openalex.publishers.publishers` is now a **static table**, edited in place. This notebook rebuilt it from the guts Postgres via the `openalex_guts` federated catalog (formerly `openalex_postgres`), which has been **removed** — the source query below can no longer run. Last state of guts `mid.publisher` (10,727 rows) matched the walden table; the walden copy additionally carries manual fixes applied 2026-01-21.

```sql
CREATE OR REPLACE TABLE openalex.publishers.publishers
AS SELECT 
    p.publisher_id as id,
    p.display_name,
    p.alternate_titles,
    p.country_code,
    CASE 
      WHEN p.parent_publisher IS NOT NULL THEN
        STRUCT(
          CONCAT('https://openalex.org/P', p.parent_publisher) as id,
          parent.display_name as display_name
        )
      ELSE NULL
    END as parent_publisher,
    p.ror_id,
    p.wikidata_id,
    p.country_name,
    p.parent_publisher_wikidata,
    p.hierarchy_level,
    p.country_codes,
    p.image_url,
    p.image_thumbnail_url,
    p.homepage_url,
    p.merge_into_id,
    p.merge_into_date,
    p.created_date,
    p.created_date as updated_date
FROM openalex_guts.mid.publisher p
LEFT JOIN openalex.publishers.publishers parent 
  ON p.parent_publisher = parent.id AND parent.merge_into_id IS NULL
```
